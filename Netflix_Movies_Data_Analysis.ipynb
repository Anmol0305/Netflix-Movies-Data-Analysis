## Dropping Unnecessary Columns

```python
cols = ['Overview', 'Original_Language', 'Poster_Url']
df.drop(cols, axis=1, inplace=True)
df.columns
```

```python
df.head()
```

## Categorizing Vote_Average Column

```python
def categorize_col(df, col, labels):
    edges = [
        df[col].describe()['min'],
        df[col].describe()['25%'],
        df[col].describe()['50%'],
        df[col].describe()['75%'],
        df[col].describe()['max']
    ]

    df[col] = pd.cut(df[col], edges, labels=labels)
    return df
```

```python
labels = ['not_popular', 'below_avg', 'average', 'popular']

categorize_col(df, 'Vote_Average', labels)
```

```python
# Check the new categories
df['Vote_Average'].unique()
```

```python
# Count each vote category
df['Vote_Average'].value_counts()
```

```python
# Remove rows with missing values
df.dropna(inplace=True)

df.isna().sum()
```

## Split Genres Into a List and Explode the DataFrame

```python
df['Genre'] = df['Genre'].str.split(', ')

df = df.explode('Genre').reset_index(drop=True)

df.head()
```

## Casting Columns Into Categories

```python
df['Genre'] = df['Genre'].astype('category')

df['Genre'].dtype
```

## Data Visualization

```python
sns.set_style('whitegrid')
```

## Q1. What is the Most Frequent Genre of Movies Released on Netflix?

```python
df['Genre'].describe()
```

```python
sns.catplot(
    y='Genre',
    data=df,
    kind='count',
    order=df['Genre'].value_counts().index
)

plt.title('Genre column distribution')
plt.show()
```

**Finding:** Drama is the most frequent genre, appearing 3,715 times in the exploded dataset.

## Q2. Which Vote-Average Category Has the Highest Count?

```python
df['Vote_Average'].value_counts()
```

```python
sns.catplot(
    y='Vote_Average',
    data=df,
    kind='count',
    order=df['Vote_Average'].value_counts().index
)

plt.title('Votes distribution')
plt.show()
```

**Finding:** The `average` category has the highest count in the exploded dataset, while the original project conclusion reports the popular category as 25.5% / 6,020 rows before genre expansion. The difference is due to genre explosion duplicating movie rows across genres.

## Q3. What Movie Has the Lowest Popularity and What Is Its Genre?

```python
df[df['Popularity'] == df['Popularity'].min()]
```

**Finding:** The minimum popularity value is 13.354. The screenshots show two movies sharing this minimum: **The United States vs. Billie Holiday** and **Threads**. Their genres are Music/Drama/History and War/Drama/Science Fiction respectively.

## Q4. Which Year Has the Most Filmed Movies?

```python
df['Release_Date'].hist()

plt.title('Release Date column distribution')
plt.show()
```

**Finding:** The project conclusion identifies **2020** as the year with the highest number of movies in the dataset.

# Conclusion

1. **Most frequent genre:** Drama.
2. **Vote popularity:** The project analysis identified the popular category as a major segment of the dataset.
3. **Lowest popularity:** The minimum popularity is 13.354; the screenshots show *The United States vs. Billie Holiday* and *Threads* at this value.
4. **Most represented release year:** 2020.
